```python
def square(x):
  return x * x

print(square(3))
```

```python
def R0(beta, gamma):
  return beta / gamma

print(R0(2.0, 1.0))
```

```python
def outbreak(beta, gamma):
  r = R0(beta, gamma)
  if r > 1:
    return "disease persists"
  else:
    return "disease dies out"

print(outbreak(2.0, 1.0))
print(outbreak(0.5, 1.0))
```

```python
for i in range(5):
  print(i)
```

```python
for beta in [0.5, 0.9, 1.0, 1.5, 3.0]:
  print(beta, outbreak(beta, 1.0))
```

```python
beta = 2.0
gamma = 1.0
h = 0.01
I = 0.10     # I(0)

for k in range(1000):
  I = I + h * (beta * I * (1 - I) - gamma * I)

print(I)
```
