```python
x = np.linspace(0, 1, 101)

u = np.cos(np.pi * x)

print(x[0])
print(x[-1])
print(len(x))
```

```python
plt.plot(x, u)
plt.xlabel("x")
plt.ylabel("u(x)")
plt.title("cos(pi x)")
plt.show()
```

```python
z = np.sin(np.pi * x)

plt.plot(x, u, label="cos")
plt.plot(x, z, label="sin")
plt.xlabel("x")
plt.ylabel("value")
plt.title("cos (pi x) and sin(pi x)")
plt.show()
```

```python
beta, gamma, h, n_steps = 2.0, 1.0, 0.01, 1000

t = np.zeros(n_steps + 1)
I = np.zeros(n_steps + 1)
I[0] = 0.1

for k in range(n_steps):
  t[k+1] = t[k] + h
  I[k+1] = I[k] + h * (beta * I[k] * (1 - I[k]) - gamma * I[k])


plt.plot(t, I)
plt.xlabel("t")
plt.ylabel("I(x)")
plt.axhline(1 - gamma/beta, linestyle="--", color="gray")
plt.show()
```

```python
gamma, h, n_steps = 1.0, 0.01, 1000

for beta in [0.8, 1.5, 2.0, 3.0]:
  t = np.zeros(n_steps + 1)
  I = np.zeros(n_steps + 1)
  I[0] = 0.1

  for k in range(n_steps):
    t[k+1] = t[k] + h
    I[k+1] = I[k] + h * (beta * I[k] * (1 - I[k]) - gamma * I[k])

  line, = plt.plot(t, I, label=f"beta = {beta}")

  if beta > gamma:
    plt.axhline(1 - gamma/beta, linestyle="--", color=line.get_color())

plt.xlabel("t")
plt.ylabel("I(t)")
plt.title("SIS model for different beta")
plt.legend()
plt.show()
```
